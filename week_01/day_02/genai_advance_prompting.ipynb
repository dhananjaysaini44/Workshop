### Import System prompts and start using

```python
ROOT = os.path.abspath("../..")

if ROOT not in sys.path:
    sys.path.append(ROOT)
```

```python
from prompts.learning import (
    LEARNING_SYSTEM_PROMPT,
    LEARNING_TEMPLATE
)
```

```python
print(LEARNING_TEMPLATE)
```

```python
prompt = LEARNING_TEMPLATE.format(
    task = "Explain embeddings and vector database"
)

print(prompt)
```

```python
response = client.models.generate_content(
    model = "gemini-3.6-flash", 
    contents = prompt,
    config = types.GenerateContentConfig(
        systemInstruction = LEARNING_SYSTEM_PROMPT,
        temperature = 0.2
    )
)

print(response.text)
```
