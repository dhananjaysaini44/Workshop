```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity #cosine_similarity requires 2D arrays
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

embedding = model.encode("My name is Dhananjay")
print(embedding.shape)
```

```python
embedding = model.encode(["What is the meaning of life?", 
                          "How does one spend time on Earth?"])

#reshape(row, column)
embed_1 = embedding[0].reshape(1, -1)
embed_2 = embedding[1].reshape(1, -1)

print(cosine_similarity(embed_1, embed_2))
```

```python
embedding = model.encode(["What is the meaning of life?", 
                          "Life is all about family and human connections"])

embed_1 = embedding[0].reshape(1, -1)
embed_2 = embedding[1].reshape(1, -1)

print(cosine_similarity(embed_1, embed_2))
```

```python
embedding = model.encode(["What is my name?", 
                          "What is my name?"])

embed_1 = embedding[0].reshape(1, -1)
embed_2 = embedding[1].reshape(1, -1)

print(cosine_similarity(embed_1, embed_2))
```

```python
words = ["King", 
         "Queen", 
         "Man", 
         "Ice-cream"]

embedding = model.encode(words)

embed_1 = embedding[0].reshape(1, -1)
embed_2 = embedding[1].reshape(1, -1)
embed_3 = embedding[2].reshape(1, -1)
embed_4 = embedding[3].reshape(1, -1)

print(cosine_similarity(embed_1, embed_2))

print(cosine_similarity(embed_1, embed_3))

print(cosine_similarity(embed_1, embed_4))
```
